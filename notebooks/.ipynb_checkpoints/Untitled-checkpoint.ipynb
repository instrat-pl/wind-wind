# Compare turbines

```python
import os
from wind_wind.paths import TURBINES_DIR

nrl_turbines = [t.replace(".yaml", "") for t in os.listdir(TURBINES_DIR / "yaml")]

nrl_turbines
```

```python
import pandas as pd
from wind_wind.paths import CUTOUTS_DIR, INPUT_DIR
from wind_wind.profiles import Wind
from wind_wind.consts import atlite_default_turbines
turbines = atlite_default_turbines + nrl_turbines
turbines = [t for t in turbines if (not "offshore" in t)]
print(turbines)
result = []
profiles = []
for turbine_name in turbines:
    stats = {}
    wind_profile = Wind(turbine_name=turbine_name, smooth=True)
    wind_profile.generate()
    stats["turbine"] = turbine_name
    stats["cf"] = wind_profile.cf
    result.append(stats)
    profile = wind_profile.profile
    profile.name = turbine_name
    profiles.append(profile)
    # wind_profile.save_to_db()
result = pd.DataFrame(result)
profiles = pd.concat(profiles, axis=1)
print(result)
```

```python
profiles[3000:3168].plot(figsize=(18,6))
```

```python
result.sort_values(by=["cf"])
```
